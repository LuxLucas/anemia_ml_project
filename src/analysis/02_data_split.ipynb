```python
import pandas as pd
from sklearn.model_selection import train_test_split
import os
```

```python
PATH = "../../data/raw/anemia_dataset.csv"
dataset = pd.read_csv(PATH)

unnecessary_columns = [
    "HGB_Anemia_Class",
    "Iron_anemia_Class",
    "Folate_anemia_class",
    "B12_Anemia_class"
]

features = (
    dataset
    .drop(columns=unnecessary_columns)
    .drop_duplicates()
)
```

```python
y = features.All_Class
X = features.drop(columns="All_Class")

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=1822,
    stratify=y
)
```

```python
PATH = "../../data/processed"
os.makedirs(PATH, exist_ok=True)

files = {
    "X_train.csv": X_train,
    "X_test.csv": X_test,
    "y_train.csv": y_train,
    "y_test.csv": y_test,
}

for filename, data in files.items():
    file_path = os.path.join(PATH, filename)
    data.to_csv(file_path, index=False)
    print(f"Salvo: {file_path}")
```
